# Data Cleaning Strategy

The datasets share a similar column structure but contain several common issues: NaN values, incorrect data types in numeric columns, duplicated columns in different languages, and dataset-specific inconsistencies.

- Keep only column with English names, as the project will be presented in English.
- Remove NaN values that are metadata-related or cannot be recovered.
- Ensure all numeric columns are converted to proper numeric data types.
- Standardize numeric values to a consistent format (e.g., adjust for cases where values are given in thousands or percentages).
- Reshape each dataset so that the year becomes the index, and values become columns.
- If regional column are present (as in economic datasets), remove them and keep only the country-level aggregate.
- Remove all data prior to 2011 to ensure consistency across datasets.

The goal is to merge all datasets into a single solid dataset. To achieve this, each dataset must first be cleaned individually using a general-purpose function that handles common issues. Dataset-specific problems will be addressed separately as they arise. The merging process will be addressed subsequently.

```python
def clean_dataset(dataset, drop_nans=True, drop_cols=None, min_year=2011, dataset_type=None, value=None):
    """
    Clean and standardize a dataset.

    - Optionally drops rows with NaN values and specified columns.
    - Reshapes data so years become the index.
    - Converts values to numeric types.
    - Removes data prior to 2011.
    - For 'economy' datasets, retains only the first, country-aggregate column and renames it.

    Parameters:
    - dataset: Input dataset.
    - drop_nans: Drop rows with NaN values.
    - drop_cols: Columns to drop.
    - min_year: Minimum year to retain. If provided less than 2011, retains only data from 2011 onward.
    - dataset_type: Dataset type (e.g., 'economy').
    - value: The name to assign to the retained column (for 'economy' datasets).

    Returns a cleaned DataFrame for further preprocessing.
    """

    clean_data = dataset.copy()

    if drop_nans:
        clean_data = clean_data.dropna()

    if drop_cols:
        clean_data = clean_data.drop(columns=drop_cols)
    
    clean_data = clean_data.set_index('Items').T
    clean_data.index = pd.to_datetime(clean_data.index, format='%Y', errors='coerce').year
    clean_data.index.name = 'Year'
    clean_data.columns.name = None

    clean_data = clean_data.apply(pd.to_numeric, errors='coerce')
    
    if min_year < 2011:
        clean_data = clean_data[clean_data.index >= 2011]

    if dataset_type == 'economy':
        clean_data = clean_data.iloc[:, :1]
        if value:
            clean_data.columns = [value]
    
    return clean_data
```

# Data Cleaning

## birth_rate

```python
clean_birth_count = clean_dataset(
    birth_count,
    drop_cols=['Көрсөткүчтөрдүн аталыштары', 'Наименование показателей'],
    dataset_type='economy',
    value='Birth Count'
)
clean_birth_count
```

The index value for 2024 was incorrectly formatted as `"2024 1"`, which caused parsing issues. As a result, it was converted to `NaT` during `to_datetime`, and the index was subsequently downcast to float when applying `to_numeric`.

Also, values represent discrete measure, so they should be integers.

```python
clean_birth_count.index = pd.to_numeric(clean_birth_count.index, errors='coerce')
clean_birth_count.index = clean_birth_count.index.fillna(2024).astype(int)
clean_birth_count.index = pd.to_datetime(clean_birth_count.index, format='%Y').year
clean_birth_count = clean_birth_count.astype(int)
clean_birth_count
```

## employed_population

```python
clean_employed_population = clean_dataset(
    employed_population,
    drop_cols=["Көрсөткүчтөрдүн аталышы", "Наименование показателей"],
    dataset_type='economy',
    value='Employed Population'
)
clean_employed_population
```

The values are expressed in *thousands of people* according to metadata. Therefore, the data should be standardized by multiplying all values by 1,000 and converting them to integers.

```python
clean_employed_population = (clean_employed_population * 1000).astype(int)
clean_employed_population
```

## gdp_per_capita

```python
clean_gdp_per_capita = clean_dataset(
    gdp_per_capita,
    drop_cols=["Көрсөткүчтөрдүн аталыштары", "Наименование показателей"],
    dataset_type='economy',
    value='GDP per Capita'
)
clean_gdp_per_capita
```

Since the metadata indicates "thousand som", the values must be multiplied by 1000.

Additionally, the function removed the column containing equivalent values in USD. This is not critical, as the analysis focuses on the local currency.

```python
clean_gdp_per_capita = clean_gdp_per_capita * 1000
clean_gdp_per_capita
```

A major issue is the absence of values for the period of 2011 to 2019. We will impute those values using data from the same source (stat.gov.kg), ensuring consistency.

Values do not contain fractional parts. Although they may be stored as `float`, they effectively represent whole numbers and should be integers.

```python
new_data = pd.Series({
    2011: 54400,
    2012: 58000,
    2013: 65000,
    2014: 71800,
    2015: 75500,
    2016: 81800,
    2017: 89300,
    2018: 88400,
    2019: 99200
})

clean_gdp_per_capita = clean_gdp_per_capita.reindex(range(2011, 2025))
clean_gdp_per_capita["GDP per Capita"] = clean_gdp_per_capita["GDP per Capita"].combine_first(new_data)
clean_gdp_per_capita = clean_gdp_per_capita.astype(int)
clean_gdp_per_capita
```
