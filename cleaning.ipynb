# Data Cleaning Strategy

After initial inspection of the files, datasets share similar column structures, contain NaN values, duplicate columns in different languages, and other issues.

- Retain only columns with English names, as the project will be presented in English.
- Remove NaN values that cannot be recovered or are meaningless, along with metadata.
- Transpose each dataset so that year columns become the index, and categories become columns.
- Standardize numeric values to a consistent format (e.g., accounting for metadata indicating values in thousands or tens of thousands).
- Address remaining problems on a case-by-case basis as they arise.

The objective is to merge all data into a single solid dataset. To achieve this, each dataset must first be cleaned individually. The merging process will be addressed subsequently.

# birth_rate

```python
birth_rate
```

```python
clean_birth_rate = birth_rate.copy()

clean_birth_rate = clean_birth_rate.drop(columns=["Көрсөткүчтөрдүн аталыштары", "Наименование показателей"])
clean_birth_rate.dropna(inplace=True) # the only row with NaN is metadata, so we can safely drop it
clean_birth_rate = clean_birth_rate.set_index('Items').T

# Convert index to datetime year and fix '2024 1' to 2024
clean_birth_rate.index = pd.to_datetime(clean_birth_rate.index.astype(str).str.replace(' 1', ''), format='%Y').year
clean_birth_rate.index.name = 'Year'

# Remove the 'Items' name from columns
clean_birth_rate.columns.name = None
```

```python
clean_birth_rate
```

# employed_population

```python
employed_population
```

```python
clean_employed_population = employed_population.copy()

clean_employed_population = clean_employed_population.drop(columns=["Көрсөткүчтөрдүн аталышы", "Наименование показателей"])
clean_employed_population.dropna(inplace=True)
clean_employed_population = clean_employed_population.set_index('Items').T

clean_employed_population.index = pd.to_datetime(clean_employed_population.index, format='%Y').year
clean_employed_population.index.name = 'Year'

clean_employed_population.columns.name = None

# Multiply by 1000 to standardize values since metadata indicates "thousand people"
clean_employed_population = clean_employed_population * 1000

# Impute missing values for 2011 and 2012 in 'Osh city' based on 2013 value minus 6k and 3k respectively, as values show an average annual growth of 3k people
clean_employed_population.loc[2011, 'Osh city'] = clean_employed_population.loc[2013, 'Osh city'] - 6000
clean_employed_population.loc[2012, 'Osh city'] = clean_employed_population.loc[2013, 'Osh city'] - 3000

```

```python
clean_employed_population
```

# gdp_per_capita

```python
gdp_per_capita
```

```python
clean_gdp_per_capita = gdp_per_capita.copy()

clean_gdp_per_capita = clean_gdp_per_capita.drop(columns=["Көрсөткүчтөрдүн аталыштары", "Наименование показателей"])
clean_gdp_per_capita.dropna(inplace=True)
clean_gdp_per_capita = clean_gdp_per_capita.set_index('Items').T

clean_gdp_per_capita.index = pd.to_datetime(clean_gdp_per_capita.index, format='%Y').year
clean_gdp_per_capita.index.name = 'Year'

clean_gdp_per_capita.columns.name = None

# Multiply by 1000 to standardize values since metadata indicates "thous. soms"
clean_gdp_per_capita["Gross domestic product per capita (thous. soms)"] = clean_gdp_per_capita['Gross domestic product per capita (thous. soms)'] * 1000
```

# unemployment

```python
unemployment
```

```python
clean_unemployment = unemployment.copy()

clean_unemployment = clean_unemployment.drop(columns=["Көрсөткүчтөрдүн аталыштары", "Наименование показателей"])
clean_unemployment.dropna(inplace=True)
clean_unemployment = clean_unemployment.set_index('Items').T

clean_unemployment.index = pd.to_datetime(clean_unemployment.index, format='%Y').year
clean_unemployment.index.name = 'Year'

clean_unemployment.columns.name = None

# Convert all columns to float, handling non-numeric values
clean_unemployment = clean_unemployment.apply(pd.to_numeric, errors='coerce')

# Convert percentage values to proportions
clean_unemployment = clean_unemployment / 100

clean_unemployment = clean_unemployment.drop(index=[2006, 2007, 2008, 2009, 2010])

# Impute missing values for Osh city based on trend
# Trend of unemployment decreasing by 0.003 each year
clean_unemployment.loc[2011, 'Osh city'] = clean_unemployment.loc[2013, 'Osh city'] + 0.006
clean_unemployment.loc[2012, 'Osh city'] = clean_unemployment.loc[2013, 'Osh city'] + 0.003
```
